### Data loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# --------------------------------------------------
# 1. LOAD DATA
# --------------------------------------------------

# Change this to your actual dataset filename

df = pd.read_csv(r"C:\Users\Admin\Desktop\Training\Python Training\25-11-2026\data\IBM-HR-Employee-Attrition.csv")


# Keep the original dataset unchanged
raw_df = df.copy()

print("Original dataset shape:", df.shape)

display(df.head())
```

### Before Summary

```python
# --------------------------------------------------
# 2. BEFORE CLEANING SUMMARY
# --------------------------------------------------

def create_summary(data):
    return pd.DataFrame({
        "Rows": [data.shape[0]],
        "Columns": [data.shape[1]],
        "Missing Values": [data.isna().sum().sum()],
        "Duplicate Rows": [data.duplicated().sum()]
    })


before_summary = create_summary(df)

print("Before Cleaning:")
display(before_summary)

print("\nMissing values by column:")
display(df.isna().sum().to_frame("Missing Values"))

print("\nData types:")
display(df.dtypes.to_frame("Data Type"))
```

### Handle Missing Values

```python
df.isna().sum()
```

```python
# --------------------------------------------------
# 6. HANDLE MISSING VALUES
# --------------------------------------------------
# Fill missing numerical values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

df["MonthlyIncome"] = df["MonthlyIncome"].fillna(df["MonthlyIncome"].median())

df["YearsAtCompany"] = df["YearsAtCompany"].fillna(df["YearsAtCompany"].median())


# Fill missing categorical values with mode
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

df["BusinessTravel"] = df["BusinessTravel"].fillna(df["BusinessTravel"].mode()[0])
```

```python
print(df.isna().sum())

print("Total missing values:", df.isna().sum().sum())
```

```python
df.head()
```

### Fix Data Types

```python
# --------------------------------------------------
# 4. FIX DATA TYPES
# --------------------------------------------------

# Numerical columns
df["Age"] = df["Age"].astype(int)
df["MonthlyIncome"] = df["MonthlyIncome"].astype(int)
df["YearsAtCompany"] = df["YearsAtCompany"].astype(int)

# Categorical columns
df["Department"] = df["Department"].astype("category")
df["BusinessTravel"] = df["BusinessTravel"].astype("category")
df["Attrition"] = df["Attrition"].astype("category")
df["EducationField"] = df["EducationField"].astype("category")
df["Gender"] = df["Gender"].astype("category")
df["JobRole"] = df["JobRole"].astype("category")
df["MaritalStatus"] = df["MaritalStatus"].astype("category")
df["Over18"] = df["Over18"].astype("category")
df["OverTime"] = df["OverTime"].astype("category")
print("Data types after conversion:")

display(df.dtypes.to_frame("Data Type"))
```

### Remove Duplicates

```python
# --------------------------------------------------
# 5. REMOVE DUPLICATES
# --------------------------------------------------

duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicates before:", duplicates_before)
print("Duplicates removed:", duplicates_before - duplicates_after)
print("Duplicates after:", duplicates_after)
```

### Outlier Handling

```python
# --------------------------------------------------
# 7. OUTLIER DETECTION AND CAPPING
# --------------------------------------------------

# Exclude ID columns and columns with binary values
# from outlier detection.

outlier_columns = []

for col in df.select_dtypes(include=["number"]).columns:

    if any(word in col.lower() for word in ["id", "code", "zip"]):
        continue

    if df[col].nunique() <= 2:
        continue

    outlier_columns.append(col)


outlier_summary = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ).sum()

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outliers Detected": outlier_count
    })

    # Cap the outliers
    df[col] = df[col].clip(
        lower=lower_limit,
        upper=upper_limit
    )


outlier_summary_df = pd.DataFrame(outlier_summary)

print("Outlier Detection Summary:")
display(outlier_summary_df)

print("Outlier treatment completed.")
```

### Before & After Summary

```python
# --------------------------------------------------
# 8. BEFORE / AFTER SUMMARY
# --------------------------------------------------

after_summary = create_summary(df)

summary_table = pd.DataFrame({
    "Before Cleaning": before_summary.iloc[0],
    "After Cleaning": after_summary.iloc[0]
})

summary_table["Change"] = (
    summary_table["After Cleaning"]
    - summary_table["Before Cleaning"]
)

print("Before and After Cleaning Summary:")
display(summary_table)

# Save cleaned dataset
df.to_csv(r"C:\Users\Admin\Desktop\Training\Python Training\25-11-2026\data\clean_data.csv", index=False)

# Save the summary table
summary_table.to_csv(
    "cleaning_summary.csv"
)

print("Cleaned dataset saved as clean_data.csv")
print("Summary saved as cleaning_summary.csv")
```
