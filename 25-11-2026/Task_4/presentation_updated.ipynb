# Employee Attrition Analysis Report

## 1. Problem Statement

The objective is to analyse employee data and understand
employee attrition patterns.



## 2. Dataset Description

The dataset contains 1470 employees and 35 columns.

237 employees have left the organisation.

Overall attrition rate: 16.12%.


## 3. Three Key Insights

1. Sales department has the highest attrition rate
   of 20.82%.

2. The 18-25 age group has the highest attrition rate.
   The 26-35 age group has the highest attrition count
   with 112 employees.

3. Employees working overtime have a 30.53% attrition
   rate compared with 10.44% for employees not
   working overtime.


## 4. Recommendations

1. Improve employee retention through career development
   and employee engagement programmes.

2. Review overtime workloads and improve work-life balance.


## 5. Limitations

- Historical employee data is used.
- Exact reasons for attrition are not available.
- Correlation does not prove causation.
- Results may not apply to other organisations.


## 6. Conclusion

The analysis helps HR understand employee attrition
patterns and identify areas for improvement.
